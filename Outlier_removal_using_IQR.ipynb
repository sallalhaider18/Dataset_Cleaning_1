```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

np.random.seed(42)

# dataset
df = pd.DataFrame({
    "age": np.random.normal(35, 10, 1000),
    "income": np.random.normal(50000, 15000, 1000),
    "spending_score": np.random.normal(50, 12, 1000),
    "height": np.random.normal(170, 10, 1000),
    "exam_score": np.random.normal(75, 10, 1000)
})

# Add real outliers
outliers = pd.DataFrame({
    "age": [120, -10],
    "income": [500000, -20000],
    "spending_score": [200, -50],
    "height": [250, 40],
    "exam_score": [220, -30]
})

df = pd.concat([df, outliers], ignore_index=True)

print(df.head())
print("Shape:", df.shape)
```

```python
plt.figure(figsize=(12,6))
df.boxplot()
plt.title("Before Removing Outliers")
plt.xticks(rotation=45)
plt.show()
```

```python
df.hist(figsize=(12,8), bins=30)
plt.suptitle("Feature Distributions")
plt.show()
```

IQR Outlier Detection

```python
def detect_outliers_iqr(data):
    outlier_indices = set()

    for col in data.select_dtypes(include=np.number).columns:
        Q1 = data[col].quantile(0.25)
        Q3 = data[col].quantile(0.75)

        IQR = Q3 - Q1

        lower = Q1 - 1.5 * IQR
        upper = Q3 + 1.5 * IQR

        outliers = data[(data[col] < lower) | (data[col] > upper)].index
        outlier_indices.update(outliers)

    return list(outlier_indices)
```

```python
outlier_idx = detect_outliers_iqr(df)

print("Total outlier rows:", len(outlier_idx))
```

```python
df_clean = df.drop(index=outlier_idx)

print("Original shape:", df.shape)
print("Clean shape:", df_clean.shape)
```

```python
plt.figure(figsize=(12,6))
df_clean.boxplot()
plt.title("After Removing Outliers (IQR)")
plt.xticks(rotation=45)
plt.show()
```

```python
df_clean.hist(figsize=(12,8), bins=30)
plt.suptitle("After Outlier Removal")
plt.show()
```
