```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from scipy import stats

np.random.seed(42)

# dataset
df = pd.DataFrame({
    "age": np.random.normal(35, 10, 1000),
    "income": np.random.normal(50000, 15000, 1000),
    "spending_score": np.random.normal(50, 12, 1000),
    "height": np.random.normal(170, 10, 1000),
    "exam_score": np.random.normal(75, 10, 1000)
})

# Add real outliers
outliers = pd.DataFrame({
    "age": [120, -10],
    "income": [500000, -20000],
    "spending_score": [200, -50],
    "height": [250, 40],
    "exam_score": [220, -30]
})

df = pd.concat([df, outliers], ignore_index=True)

print(df.head())
print("Shape:", df.shape)
```

Z-score Outlier Detection

```python
def detect_outliers_zscore(data, threshold=3):
    z_scores = np.abs(stats.zscore(data.select_dtypes(include=np.number)))
    outlier_rows = np.where(z_scores > threshold)[0]
    return np.unique(outlier_rows)
```

```python
z_outliers = detect_outliers_zscore(df)

print("Z-score outliers:", len(z_outliers))
```

```python
df_z_clean = df.drop(index=z_outliers)

print("Shape after Z-score cleaning:", df_z_clean.shape)
```
